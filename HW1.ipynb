```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#read the data
df = pd.read_csv('sampregdata.csv')
df.head()
```

```python
#since I don't need unnamed column, I will drop it
df = df.drop(columns=['Unnamed: 0'], errors='ignore')

#define the target and features - for only one feature
target = 'y'
features = 'x1'
results = []

#split the data into train and test sets
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
X_train = train_df[['x1']] #double brackets to make it a dataframe. This is the strict requirement of sklearn
y_train = train_df[target]
X_test = test_df[['x1']]
y_test = test_df[target]

model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

#use r2_score and mean_squared_error to evaluate the model
r2 = r2_score(y_test, predictions)
mse = mean_squared_error(y_test, predictions)

#dictionary to store the results
results.append({
    'model': 'LinearRegression',
    'features': features,
    'targeet': target,
    'r2_score': r2,
    'mean_squared_error': mse
})

results_df = pd.DataFrame(results)
print(results_df)
```

```python
#use the for loop to iterate through the features and train a model for each feature
target = 'y'
features = ['x1', 'x2', 'x3','x4']
#split the data into train and test sets
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
results = []
for feature in features:
    X_train = train_df[[feature]] 
    y_train = train_df[target]
    X_test = test_df[[feature]]
    y_test = test_df[target]

    model = LinearRegression()
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    #use r2_score and mean_squared_error to evaluate the model
    r2 = r2_score(y_test, predictions)
    mse = mean_squared_error(y_test, predictions)

    #dictionary to store the results
    results.append({
        'features': feature,
        'r2_score': r2,
        'mean_squared_error': mse
        })
results_df = pd.DataFrame(results)
print("linear Regression:\n", results_df)
```

```python
#2X model
#define the target and features
target = 'y'
features = ['x1', 'x2', 'x3','x4']
#split the data into train and test sets
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
results = []

#pair the possible combinations of features
from itertools import combinations
feature_cominations = combinations(features, 2)
for feature in feature_cominations:
    features = list(feature)
    X_train = train_df[features]
    y_train = train_df[target]
    X_test = test_df[features]
    y_test = test_df[target]

    model = LinearRegression()
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    #use r2_score and mean_squared_error to evaluate the model
    r2 = r2_score(y_test, predictions)
    mse = mean_squared_error(y_test, predictions)

    #dictionary to store the results
    results.append({
        'features': features,
        'r2_score': r2,
        'mean_squared_error': mse
        })

results_df = pd.DataFrame(results)
print("linear Regression with 2 features:\n", results_df)
```

## R-squared
It measures how much of the variation in the target is explained by the model, which means to know the performance of the model.
## Mean Squared Error
It measures the average squared difference between the actual values and predicted values.
